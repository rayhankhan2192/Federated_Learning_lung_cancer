```python
import os
import shutil
import random

base_dir = r'E:\luung data\niaz data\Lung-CT Scan-20250616T211158Z-1-001\Lung-CT Scan'
classes = ['Normal cases', 'Bengin cases', 'Malignant cases']
target_count = 1000

for cls in classes:
    folder = os.path.join(base_dir, cls)
    images = os.listdir(folder)
    current_count = len(images)
    print(f"{cls}: {current_count} images")

    if current_count < target_count:
        needed = target_count - current_count
        for i in range(needed):
            src_img = random.choice(images)
            # Create a new filename to avoid overwrite
            name, ext = os.path.splitext(src_img)
            new_filename = f"{name}_dup_{i}{ext}"
            shutil.copy(os.path.join(folder, src_img), os.path.join(folder, new_filename))
        print(f"{cls}: Now has {len(os.listdir(folder))} images (oversampled to {target_count})")
    elif current_count > target_count:
        print(f"Warning: {cls} already has more than {target_count} images. Consider removing extras.")
    else:
        print(f"{cls}: Already has {target_count} images")

print("✅ All class folders now have 1000 images.")
```
